#Match analysis using numpy
The point of this notebook is to practice using numpy.

I create fake data and analyze it using numpy functions such as mean, max, masking, etc.

I know this is not all the data a football match has. This is just practice

##Initialization

```python
import numpy as np
```

```python
# creating a random number generator
rng = np.random.default_rng(seed=23)
```

```python
# creating arrays of goals scored, goals conceded, shots, and possesion
goalsScored = rng.integers(0, 6, size=100)
goalsConceded = rng.integers(0, 6, size=100)
shots = rng.integers(5, 20, size=100)
possesion = np.around(rng.uniform(30, 70, size=100), 2)
```

```python
# stacking the arrays into one 100x4 array
matches = np.column_stack((shots, goalsScored, goalsConceded, possesion))
matches
```

## Analysis

```python
# computing the data analysis
avgGoalsScored = np.mean(matches[:,1])
avgGoalsConceded = np.mean(matches[:, 2])
avgShots = np.mean(matches[:,0])
maxShots = np.max(matches[:,0])
minShots = np.min(matches[:, 0])
goalDifference =  matches[:,1] - matches[:, 2]
```

```python
# adding goal difference to matches
matches = np.column_stack((matches, goalDifference))
```

```python
# matches where possession was greater than or equal to 50%
matchesFifty = matches[:,3] >= 50.0
nMatchesFifty = np.count_nonzero(matchesFifty)
```

```python
nMatchesFifty
```

```python
# matches where goal difference is more than 1
diffOne = np.abs(matches[:,1] - matches[:,2]) > 1
ndiffOne = np.count_nonzero(diffOne)
```

```python
# matches where number of shots is higher than the average
moreShotAvg =  matches[:, 0] > avgShots
nShotAvg = np.count_nonzero(moreShotAvg)
```

```python
moreShotAvg
```

```python
nShotAvg
```

```python
# making results array (-1 = loss, 0 = draw, 1 = win)
result = np.sign(goalDifference)
result
```

```python
# adding results to matches
matches = np.column_stack((matches, result))
```
